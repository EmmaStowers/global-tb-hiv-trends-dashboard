**Abstract**  
&nbsp;&nbsp;&nbsp;&nbsp;This project examines the global evolution of the tuberculosis (TB) burden during a pivotal era of international health policy. Utilizing World Health Organization (WHO) data spanning 1990 to 2013, the analysis explores the correlation between TB incidence and HIV prevalence across six global regions. TB is a highly contagious lung disease that remains one of the most fatal infectious diseases worldwide, despite being preventable and treatable. It particularly affects low- and middle-income countries, and is closely linked with HIV infection, health infrastructure quality, and socioeconomic inequality (WHO, 2026). By focusing on 1990 to 2013, this project evaluates whether international health policies corresponded with measurable improvements in TB incidence, mortality, and case detection worldwide, while also allowing us to see regions where progress was uneven.


**Project Introduction & Research Questions**  
&nbsp;&nbsp;&nbsp;&nbsp;This project focuses on the global patterns of tuberculosis (TB) incidence, prevalence, and mortality, relating to regional differences and the interaction between TB and Human Immunodeficiency Virus (HIV). The timeframe of 1990 to 2013 allows analysis of the disease in a time when TB was declared a global emergency by the World Health Organization (WHO, 2013). During this time, the Directly Observed Therapy (DOT) for tuberculosis treatment strategy was implemented worldwide to ensure that TB patients adhere to their treatment regimens (CDC, n.d.). Meanwhile, large-scale funding initiatives expanded access to diagnosis and treatment. The early 2000s were characterized by an increase in funding and cooperation against infectious diseases, such as with the Global Fund to Fight AIDS, Tuberculosis, and Malaria, a funding program supported by public governments to mitigate the impact of these diseases in countries in need (The Global Fund, n.d.). In addition, around this same time, the WHO declared HIV/AIDS pandemic a global heath emergency, leading to the increased access to antiretroviral therapy (ART) in developing countries(WHO, 2003). As a result, the dataset captures both the height of global TB burden and the effects of coordinated intervention. What initially motivated me to choose this topic was my interest in biology, as well as my experience taking a class on infectious diseases in which two of the major focuses were HIV and TB. This gave me some background knowledge when interpreting the data set, and sparked my interest in the topic.  

&nbsp;&nbsp;&nbsp;&nbsp;The guiding questions for this analysis focus on uncovering the most meaningful trends within the dataset. Specifically, I examine how TB incidence and mortality rates changed over time, both globally and regionally, to understand the overall trajectory of the disease. I also analyze which regions experienced the most significant improvements in case detection and disease control, allowing for a clearer evaluation of how effective global health interventions were between 1990 and 2013. Finally, I explore the relationship between HIV prevalence among TB cases and TB mortality, since co-infection was a major factor influencing outcomes during this period.
  
*Research Questions:*
- How did TB incidence and mortality rates change globally and regionally between 1990 and 2013?
- Which regions experienced the most significant improvements in case detection and disease control between 1990 and 2013?
- What relationship exists between HIV prevalence among TB cases and TB mortality rates during this period?


**Data Description**  
Data Source: [Tuberculosis Burden by Country](https://public.tableau.com/app/sample-data/TB_Burden_Country.csv)   
- Estimates the prevalence and mortality of Tuberculosis by country. Source: [World Health Organization](https://www.who.int/data/)
- Rows: 5121
- Columns: 47

&nbsp;&nbsp;&nbsp;&nbsp;The data set provides 24 years of standardized, country-level data that can be used for regional comparisons. It contains data on TB incidence, mortality, and case detections, as well as HIV mortality, incidence, and coinfection.  Because these measures are provided both as normalized rates (per 100,000 population) and absolute counts, the data allows for standardized comparison across countries. The data set contains numerous categorical variables, such as country name, region, and ISO codes, as well as temporal data in the form of years. In addition to this, the majority of the data is quantitative relating to incidence, mortality, case detection, and HIV coinfection.  

&nbsp;&nbsp;&nbsp;&nbsp;The most significant limitation in the dataset is that most variables are estimated. The WHO does not count every TB case; instead, they use mathematical models to fill in the gaps. In order for me to interpret my data, I had to make the assumption that the WHO accurately estimated its data with a relatively small margin of error. In regions such as Africa or Southeast Asia, the data could be subject to the bias of the algorithms used by the WHO, potentially making the estimates inaccurate. The potential for a surveillance bias also exists– as a country’s health system improves, it finds more cases, making the incidence seem like it's increasing when the country is really just improving its case detection. When calculating regional averages, nulls can also potentially skew results. For example, if the countries with the highest burden do not report their data, the results for the overall region may appear lower than they are in reality. This data set was mostly complete, so the largest concern relating to this issue is in the HIV coinfection data. Overall, this data set was very clean and did not pose many limitations




**Data Cleaning & Preprocessing**  
&nbsp;&nbsp;&nbsp;&nbsp;In terms of data cleaning, the dataset required relatively minimal preprocessing due to its overall structure and consistency. Variables were already stored in appropriate data types, there were no duplicate records, and naming conventions were largely standardized across columns and rows. The majority of countries had complete data for TB prevalence and mortality, and the data appeared to be within a reasonable range for all countries based on population. However, variables related to HIV prevalence, HIV-related deaths, and TB-HIV co-infection contained more missing values. Rather than imputing these values, which could introduce bias, I allowed Tableau to exclude null entries by default. Given the dataset’s size and coverage, there was sufficient data to garner meaningful trends. To further mitigate the impact of missing data, most visualizations aggregate results at the regional level. This aggregation reduces sensitivity to country-level gaps and allows broader patterns to remain visible, while missing data in the global map is explicitly represented by leaving countries unshaded in affected years, maintaining transparency in data availability. Overall, the majority of the data cleaning came from the need to reduce the 47 columns into something easier to manage, but even that was not fully necessary. I originally cleaned the data with the goal of reducing complexity in the number of variables, but after constructing a few visuals, I realized where I could improve on what data to keep, remove, and rename. At that point, I recleaned the data and imported the newly cleaned set later on.  

&nbsp;&nbsp;&nbsp;&nbsp;In terms of data transformation, there was not much that needed to be done due to the fact that ​​the dataset was already standardized. All of the countries already had their column variables in the format of “rate per 100k population” in order to allow fair comparison among countries of varying sizes. Additionally, the dataset included population values and percentage-based measures, making calculations for standardization extremely straightforward. The primary transformation involved aggregating data by region and year.  With the sheer number of data points provided in the set, it would have been extremely cluttered to include every individual country, and obscured broader patterns. Regional aggregation not only improved interpretability but also aligned more directly with the project’s goal of identifying global and regional trends. In addition to these transformations, I also created a few new calculated fields, such as “TB Incidence HIV Neg” and “Total TB Mortality,” to improve some visuals and enhance the dataset’s analytical value without altering its underlying structure.  

&nbsp;&nbsp;&nbsp;&nbsp;Many of these design decisions were built on the data preparation and transformation choices discussed earlier, with the overall goal of making the visualizations clearer and more meaningful. When creating the region aggregations, I decided to remain with the World Health Organization's regions, despite some being considered uncommon, to ensure the scientific validity and consistency of regional comparisons. Additionally, I consistently used rates “per 100,000 population” rather than total case counts so that large countries did not completely overshadow the smaller ones.


**Exploratory Data Analysis**   
&nbsp;&nbsp;&nbsp;&nbsp;The initial phase of the analysis involved a statistical overview of the dataset, which comprises 5,120 records representing 219 countries and territories from 1990 to 2013. There is significant variation in the data, with some countries reporting near-zero incidence while others report over 1,000 cases per 100,000, indicating extreme regional disparities and geographic inequality.  

&nbsp;&nbsp;&nbsp;&nbsp;In order to see the general distribution of the disease, I created initial visualizations focusing on regional burden and basic incidence trends. I created a choropleth map to visualize TB incidence rates per 100,000. This immediately highlighted regional disparities, particularly with heavy disease burden being concentrated in Africa and South East Asia. I also created a box plot in order to see the general regional trends in how dispersed the TB incidence per 100,000 was among the regions, indicating the largest improvements occurred in Africa, South East Asia, and the Western Pacific region. In contrast, the other three regions started with significantly lower incidence and saw little improvement over the 24-year period. In addition to this, I created a dot plot of HIV incidence in TB compared to total incidence and layered a trend line over the data in order to see that there is a strong positive correlation between the two variables.   

&nbsp;&nbsp;&nbsp;&nbsp;The main observations made during the exploratory phase were that HIV and TB incidence appear to be positively correlated. In Africa in particular, the epidemic is characterized by very high incidence rates that are more directly mirrored by high HIV prevalence. I also observed that the raw case numbers seem to be misleading. For instance, large nations in SE Asia show massive total numbers, but when normalized to 'per 100,000' people, many smaller African nations show a much more severe public health crisis. I also observed a general downward trend in TB mortality and incidence in my initial observations, and a more extreme improvement in regions with more developing countries. Overall, the initial exploration helped shape the project's research questions, particularly the importance of regional disparities, HIV-TB interaction, and the shift in mortality and incidence.

**Key Insights & Findings**  
&nbsp;&nbsp;&nbsp;&nbsp;This project told the story of how global tuberculosis (TB) burden evolved during a critical era of international public health intervention from 1990 to 2013. The data starts in 1990, when TB rates were high in many regions, and the HIV epidemic was intensifying TB-related deaths. Following 1993, when TB was declared a global health emergency by the World Health Organization, coordinated strategies such as standardized treatment protocols and expanded international funding aimed to reshape global health efforts. The data reveals that while these interventions struggled to stop the spread of new cases immediately, they were profoundly successful in saving lives.  

&nbsp;&nbsp;&nbsp;&nbsp;TB incidence and mortality rates changed significantly between 1990 and 2013. While incidence remained relatively stable overall, mortality decreased by nearly half, indicating substantial improvements in treatment outcomes. Africa and Southeast Asia appeared to be disproportionately affected by the TB epidemic, starting at a much greater disadvantage in terms of their mortality and incidence numbers. In contrast, regions such as the Americas and Europe maintained lower overall burdens and experienced more consistent declines over time. Although incidence trends remained fairly stable across most regions – with the exception of Africa – mortality declined sharply after 1993. This decline suggests that improvements in treatment adherence had a significant life-saving impact, even without large reductions in new cases. There was also a noticeable spike around 2002 in both mortality and incidence, which is likely due to improved surveillance as global funding increased thanks to the establishment of the Global Fund. While it may look like TB suddenly worsened, this change actually reflects progress in finding unidentified cases.  

&nbsp;&nbsp;&nbsp;&nbsp;The regions that experienced the most significant improvements in case detection and disease control were Southeast Asia and Africa, which is reflected in dramatic reductions in mortality. Starting in 1990, five out of six regions were below the WHO goal of 70% detection. In 2013, this changed dramatically, with only one out of six regions below this goal. Southeast Asia showed the largest improvement, increasing from about 40% to over 70% case detection, with Africa close behind, yet still below the 70% goal. These improvements were critical because the countries were able to identify and treat more cases in order to help mitigate mortality and spread. Southeast Asia and Africa also experienced the two highest reductions in TB mortality between 1990 and 2013, reflecting their stronger health systems, better diagnostic tools, and expanded global support.   

&nbsp;&nbsp;&nbsp;&nbsp;The relationship between HIV prevalence among TB cases and TB mortality rates shows a clear positive association during this period. From 1990 to approximately 2002, TB cases linked to HIV nearly tripled, and across all regions, increases in HIV prevalence correspond with rising TB mortality and incidence. HIV weakens the immune system, making individuals far more susceptible to TB. Even though TB treatments were improving, the rapid increase in HIV infections offset much of that progress, helping explain why incidence and mortality remained high in some regions despite improvements in TB treatment and detection. Around the early 2000s, there was a turning point that coincided with the expansion of global funding initiatives and increased access to antiretroviral therapy in developing countries, leading to an overall decrease in TB mortality and incidence. As HIV treatment became more widely attainable, both HIV-TB infection rates and mortality began to decline, demonstrating that meaningful progress requires coordinated global health efforts. 




**Limitations & Future Work**   
&nbsp;&nbsp;&nbsp;&nbsp;One of the primary limitations of the *Tuberculosis Burden by Country* dataset is its restricted time frame. Because the data spans only from 1990 to 2013, it is highly effective for historical analysis of major global health interventions during that period, but it does not account for more recent developments in TB control, evolving health policies, or disruptions caused by modern public health crises such as COVID-19. This limits the project’s ability to assess whether the trends observed during this period were sustained over the long term. In addition to this, the majority of countries have complete data for TB prevalence and mortality, and the data appears to be within a reasonable range for all countries based on population. However, HIV infection and deaths, along with TB and HIV coinfection and deaths, have more data missing. Although these gaps are not significant enough to affect the interpretability of overall trends, they may reduce accuracy in some regional comparisons. Additionally, many of the records are based on "predicted" or "indirect" methods, creating a degree of uncertainty compared to direct clinical surveillance.   

&nbsp;&nbsp;&nbsp;&nbsp;Future improvements could substantially expand the analytical depth of this project. Incorporating the WHO Global TB Reports from 2014 to 2024 would allow for the evaluation of the long-term sustainability of the mortality and incidence trends, and the influence of newer interventions. It may also be interesting to layer in a data set regarding COVID-19 and how it affected tuberculosis trends, similar to the current data set with HIV. Additionally, socioeconomic indicators could also be included, such as GDP per capita, healthcare spending, or the Gini coefficient, in order to provide a deeper analysis of the social determinants and disparities that affect the progression of TB. These additions would allow for a deeper analysis of how public health policy, economic development, and social inequality shape global TB trends.


**Conclusion**  
&nbsp;&nbsp;&nbsp;&nbsp;This period of 1990 to 2013 saw overall success in the treatment and control of Tuberculosis. While incidence remained relatively flat throughout this period, mortality was halved. The data appeared to show steady improvements post 1993, with the exception of HIV driven incidents and mortality trends. Post 2002, overall trends in both TB and HIV began to decline significantly. Based on research, this was likely due to the increased Global Health efforts that categorize this period, such as the 1993 WHO Declaration of TB as a global emergency and implementation of the DOTS TB treatment strategy, as well as the 2002 establishment of the Global Fund and the subsequent increase in access to antiretroviral therapy, an HIV treatment, in developing regions. At the same time, the findings emphasize that global health progress is rarely uniform.  Regions such as Africa and Southeast Asia faced disproportionately higher burdens, likely due to their historically weaker healthcare infrastructure and the compounding effects of HIV. This highlights the need for international health strategies to tailor to regional needs rather than relying solely on broad universal approaches. The research also shows the strong relationship between  HIV prevalence and TB outcomes. This demonstrates how successful disease control often requires integrated approaches that consider overlapping epidemics and broader social determinants.  

&nbsp;&nbsp;&nbsp;&nbsp;The most significant implication is that isolated healthcare can hinder improvement in patient care. The success seen after 2002 indicates that the greater success in TB when treated in tandem with HIV. These historical trends can serve as a case study for current and future global health policy. The lessons learned from TB control between 1990 and 2013 are highly relevant to modern challenges, including pandemic preparedness, healthcare equity, and sustainable international funding. Understanding what succeeded can inform responses to contemporary public health threats such as COVID-19, antimicrobial resistance, and emerging infectious diseases. Ultimately, this research matters because it reinforces that global collaboration and equitable resource distribution are essential for controlling infectious diseases at scale. It shows that while major progress is possible, long-term success depends on addressing both medical and structural inequalities worldwide.
